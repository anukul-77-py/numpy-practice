```python
import numpy as np
```

```python
#indexing of 1d array
arr=np.array([1,2,3])
print (arr)
```

```python
print(arr[0])
```

```python
print(arr[-1])
```

```python
arr[0]=5
print(arr[0])
```

```python
for i in range(0,3):
    print(arr[i])
```

```python
#slicing of 1d array
#syntax : arr[start: stop :skip ]
print(arr[0:2])#in slicing it excluded the last index value 
```

```python
print(arr[-1:])
```

```python
print(arr[-2:])
```

```python
arr_1=np.array([1,2,3,4])
```

```python
print(arr_1[-3:2])
```

```python
#indexing in 2d array
arr_2=np.array([[1,2,3],[4,5,6]])
print(arr_2)
```

```python
for i in range(0,2):
    for j in range(0,3):
        print(arr_2[i][j])
              
```

```python
#slicing in 2d array
print(arr_2[:,2])#it gives third column value 
```

```python
print(arr_2[1],arr_2[1:])
```

```python
print(arr_2[1:])
```
