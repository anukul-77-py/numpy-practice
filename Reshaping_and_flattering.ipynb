```python
#Reshaping and flattening 
# reshaping  means change matrix row and column 
#flattering means change multiple dimension into 1d(dimension)
```

```python
import numpy as np
```

```python
arr=np.array([1,2,3,4,5,6])
```

```python
print(arr)
```

```python
arr_2=arr.reshape(6,1)
print(arr_2)
```

```python
arr_3=arr.reshape(2,3)#in parameter we pass matrix size according to the number of element
print(arr_2)
```

```python
print(arr_3.flatten())#changing multiple dimension array into single(original) form
```
