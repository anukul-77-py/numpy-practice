```python
#stacking means adding two or more matrix 
# we can add vertically and horizontally according to our need
```

```python
import numpy as np
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
```

```python
print(np.vstack((a,b)))#it add vetically and change one dimension array into 2d 
```

```python
print(np.hstack((a,b)))#it add horizontally 
```

```python
c=np.array([[8,9,10],[7,5,6]])
print(c)
```

```python
hsplit=np.hsplit(c,3)#it horizontally split the matrix
print(hsplit)
```

```python
for s in hsplit:
    print(s)
    
```

```python
vsplit=np.vsplit(c, 2)#it vertically split the matrix
print(vsplit)
```

```python
for i in vsplit:
    print(i)
```
