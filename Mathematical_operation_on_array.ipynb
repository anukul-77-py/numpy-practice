```python
#Mathematical operation on Array
```

```python
import numpy as np
```

```python
a=np.array([10,20,-50])
print(a)
```

```python
print(a+10)#addition,  -it add all element of array
print(a-2)#subtraction , it subtract all elements of arrary
print(a*2)#mulitplication
print(a/10)#division
```

```python
b=np.array([1,4,9])
print(np.square(b))
print(np.sqrt(b))
```

```python
c=np.array([30,45])
print(np.cos(c))
```

```python
#Mathematical operation on multiple array
```

```python
x=np.array([1,2,3])
y=np.array([4,5,6])
print(x)
print(y)
```

```python
print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
print(np.divide(a,b))
```

```python
print(np.dot(x,y))#dot product 
```

```python
transpose1=np.array([[1,2,3],
                  [4,5,6]])
print(transpose1)
print(transpose1.T)#transpose - change row into column and column into row
```

```python
#statistical function
```

```python
st=np.array([[1,2,4],[5,6,7]])
```

```python

print(np.sum(st))#sum of all element
print(np.mean(st))#average value
print(np.median(st))#middle value
print(np.max(st))#maximum value
print(np.min(st))#minimum
print(np.std(st))#standard deviation

```
