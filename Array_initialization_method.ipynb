```python
import numpy as np
```

```python
#zeros array
zero_arr=np.zeros((2,3))#in parameter we give size of matix (row , column)
print(zero_arr)
```

```python
#ones  array
ones_arr=np.ones((2,3))
print(ones_arr)
```

```python
#full array
full_arr=np.full((2,3),7)#in full array we pass three values for size  and what number we want to print
print(full_arr)
```

```python
#identity matrix
id_arr=np.eye(3)
print(id_arr)
```

```python
#empty arrary
print(np.empty(2))#it is an empty array but it gives random value 
```

```python
#evenly spaced array
print(np.arange(1,10,2))# the first parameter is for initialize second for stop point and third one is for skip means it print 1 from 10 but prinitng skipping by 2 number
```

```python
#specific number equally between a range
print(np.linspace(1,10,4))
```

```python
#random values array for float
r_array=np.random.rand(2,3)
print(r_array)
```

```python
#random vlaues array for int
r_array=np.random.randint(1,10,(2,2)) #in this we pass three values first is range and another is size
print(r_array)
```
