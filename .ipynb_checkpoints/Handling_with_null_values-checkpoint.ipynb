```python
#Handling with null value

```

```python
import numpy as np
```

```python
a=np.array([1,3,np.nan,5,np.inf])#nan means null value and inf means inifinte value
print(a)
```

```python
print(np.isnan(a).sum())
```

```python
a=np.nan_to_num(a)
```

```python
print(a)
```
