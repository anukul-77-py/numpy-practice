```python
import numpy as np
```

```python
arr=np.array([[1,2,4],[4,6,7]])
print(arr)
```

```python
print(arr.shape)#it gives size of each dimension like a matrix (row , column)
```

```python
print(arr.size)#it gives total number of element in array
```

```python
print(arr.dtype)#it gives data type of array element
```

```python
print(arr.ndim)#it gives the number of dimensioin
```
